```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.ticker import PercentFormatter

def plot_sublabels_prob(dataset,data_name):
    bins = [0, 0.25, 0.5, 0.75, 1]
    hist_data = {}
    
    for column in ["irony_probs","toxicity_probs","humor_probs"]:
        # if column in ["text", "offensive_probs","empathy_probs","politeness_probs"]:
        #     continue
        hist, _ = np.histogram(dataset[column], bins=bins)
        # Convert counts to percentages
        hist_percentage = hist / hist.sum() * 100
        hist_data[column] = hist_percentage

    # Convert the histogram data to a DataFrame for easier plotting
    hist_df = pd.DataFrame(hist_data, index=['0-0.25', '0.25-0.5', '0.5-0.75', '0.75-1'])

    # Define the positions for the bars
    bar_width = 0.15
    bin_labels = hist_df.index
    x = np.arange(len(bin_labels))

    # Create a figure and a set of subplots
    fig, ax = plt.subplots(figsize=(10, 6))
    colors = ['#1f77b4', '#ff7f0e','#2ca02c'  ] 
    # Plot each column's histogram as bars
    for i, column in enumerate(hist_df.columns):
        bars = ax.bar(x + i * bar_width, hist_df[column], width=bar_width, label=column[:-6], color=colors[i % len(colors)])

        # Add the percentage labels on top of each bar
        for bar in bars:
            height = bar.get_height()
            ax.annotate(f'{round(height)}%',
                        xy=(bar.get_x() + bar.get_width() / 2, height),
                        xytext=(0, 2),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom',fontsize=12)
            
    # Set the x ticks to the middle of the grouped bars
    ax.set_xticks(x + 1.5 * bar_width)
    ax.set_xticklabels(bin_labels)

    ax.tick_params(axis='x', labelsize=16)
    ax.tick_params(axis='y', labelsize=16)

    # Set the y-axis limits to 0-100%
    ax.set_ylim(0, 100)

    # Add labels and title
    ax.set_xlabel('Score intervals',fontsize=18)
    ax.set_ylabel('Percentage',fontsize=18)
    # ax.set_title(f"{data_name} Sarcastic Dataset's Sublabels Probability")
    ax.legend(loc='upper center',fontsize=16)
    # Show the plot
    plt.show()
    
```

```python
# Compute correlations and p-values between F1 Score and each KL Divergence value
#'offensive_probs', 'empathy_probs', 'irony_probs', 'toxcity_probs', 'humor_probs', 'politeness_probs'

def corr_p_val_plot(data_prop):
    f1_scores =[]
    sub_labels_kl = []
    sub_labels_kl_humor = []
    sub_labels_kl_irony = []
    sub_labels_kl_politeness = []
    sub_labels_kl_toxicity = []
    sub_labels_kl_empathy = []
    sub_labels_kl_offensive = []
    
    for data_pair in data_prop.columns:
        f1_scores.append(data_prop[data_pair][6])
        sub_labels_kl_humor.append(data_prop[data_pair][0])
        sub_labels_kl_irony.append(data_prop[data_pair][1])
        sub_labels_kl_politeness.append(data_prop[data_pair][2])
        sub_labels_kl_toxicity.append(data_prop[data_pair][3])
        sub_labels_kl_empathy.append(data_prop[data_pair][4])
        sub_labels_kl_offensive.append(data_prop[data_pair][5])
        for kl in data_prop[data_pair][0:5]:
            sub_labels_kl.append(kl)
            
    print("humor : " ,pearsonr(f1_scores, sub_labels_kl_humor))
    print("irony : " ,pearsonr(f1_scores, sub_labels_kl_irony))
    print("politeness : " ,pearsonr(f1_scores, sub_labels_kl_politeness))
    print("toxicity : " ,pearsonr(f1_scores, sub_labels_kl_toxicity))
    print("empathy : " ,pearsonr(f1_scores, sub_labels_kl_empathy))
    print("offensive : " ,pearsonr(f1_scores, sub_labels_kl_offensive))
    
    colors = ['blue', 'green', 'red', 'purple','yellow','orange']

    # Plotting the data
    plt.figure(figsize=(10, 6))
    
    last_row = data_prop.iloc[-1]
    labels = [ 'humor','irony', 'politeness', 'toxicity', 'empathy', 'offensive']
    for idx in range(len(data_prop) - 1):
        x_values = data_prop.iloc[idx]
        y_values = last_row
        plt.scatter(x_values, y_values, color=colors[idx], label=labels[idx])
            

    # Titles and labels
    plt.title("Sublabels' KL Divergence and F1 Score of In and Cross Domain")
    plt.xlabel('KL divergence')
    plt.ylabel('F1 score (the gap between in domain and cross domain)')
    plt.legend()

    plt.show()

def kl_div_f1(train_data_prob,test_data_prob,in_A_A,cross_A_B):
    epsilon = 1e-10
    sublabel_kl = []
    for idx in range(6):
        p = np.array(test_data_prob[idx])
        q = np.array(train_data_prob[idx])
        q = np.where(q == 0, epsilon, q) # avoid division with zero
        sublabel_kl.append(np.sum(rel_entr(p,q)))
    f1_score = f1_score_eval(in_A_A,cross_A_B)
    sublabel_kl.append(f1_score)
    return sublabel_kl

def f1_score_eval(in_A_A,cross_A_B):
    return in_A_A-cross_A_B
```
